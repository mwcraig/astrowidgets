```python
symbol = "Security 1"
symbol2 = "Security 2"
price_data = pd.DataFrame(
    np.cumsum(np.random.randn(150, 2).dot([[0.5, 0.4], [0.4, 1.0]]), axis=0) + 100,
    columns=["Security 1", "Security 2"],
    index=pd.date_range(start="01-01-2007", periods=150),
)

dates_actual = price_data.index.values
prices = price_data[symbol].values
```

```python
sec2_data = price_data[symbol2].values


xs_pz = DateScale(min=np.datetime64(py_dtime(2007, 1, 1)))
ys_pz = LinearScale()
line_pz = Lines(
    x=dates_actual, y=sec2_data, scales={"x": xs_pz, "y": ys_pz}, colors=["red"]
)

panzoom = PanZoom(scales={"x": [xs_pz], "y": [ys_pz]})
xax = Axis(scale=xs_pz, label="Date", grids="off")
yax = Axis(scale=ys_pz, label="Price", orientation="vertical", grid_lines="none")

Figure(marks=[line_pz], axes=[xax, yax], interaction=panzoom)
```

```python

sc_x = LinearScale(min=98, max=104)
sc_y = LinearScale(min=98, max=104)
i_sc_x = LinearScale(min=98, max=104)
i_sc_y = LinearScale(min=98, max=104)
sc_size = LinearScale()

pri = prices[:20]
pri = pri - pri.min() + 98
sec2 = sec2_data[:20]
sec2 = sec2 - sec2.min() + 98
size = np.random.random_integers(1,2, size=len(pri))
size = 10 ** size
size_data = 0 * size + 0.5

scatt = Scatter(x=pri, y=sec2, scales={"x": sc_x, "y": sc_y, "size": sc_size}, size=size_data, default_size=100, fill=False)
scatt2 = Scatter(x=pri, y=sec2, scales={"x": sc_x, "y": sc_y, "size": sc_size}, size=None, default_size=100, colors=["red"], opacities=[0.4])

sc_xax = Axis(label=(symbol), scale=sc_x)
sc_yax = Axis(label=(symbol2), scale=sc_y, orientation="vertical")

panzoom2 = PanZoom(scales={"x": [sc_x], "y": [sc_y],})
f = Figure(marks=[scatt, scatt2], axes=[sc_xax, sc_yax], interaction=panzoom2)
f
```

```python
i_wid_x = i_sc_x.max - i_sc_x.min
i_wid_y = i_sc_y.max - i_sc_y.min
o_default = scatt.default_size
flooba = 0
def set_sizes(change):
    global flooba
    #print(f"{change.keys()}")
    #print(f"{change['new']=} {change['old']=} {flooba=} {sc_x.min=}")
    
    flooba = flooba + 1
    if change['new'] == change['old']:
        print("pan")
    else:
        # scale based on width of scale currently
        scale = (sc_x.max - sc_x.min) / i_wid_x # change['new'] / change['old']
        
        # scale based on idea that .size tells you the sdesired size in the data coordinates 
        scale = 1 / (scatt.size[0] / (sc_x.max - sc_x.min) * 500 )
        print(scale)
        #flooba *= scale
        
        if flooba > 0:
            scatt.default_size = max(int(o_default / scale), 1) # int(change['new'] / change['old'])
            flooba = 0
            print("-----> BOOM <------")
```

```python
sc_x.observe(set_sizes, names='max')
```
